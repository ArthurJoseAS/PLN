```python
## Utilizando uma métrica de tempo
import time

### Importing Dataset
dataset = pd.read_csv('Imdb_BOW.csv', encoding='utf-8')

X = dataset.iloc[:, :-1]
y = dataset["label"]

## Carregando o algoritmo
from sklearn.neural_network import MLPClassifier
from sklearn import metrics

# Separando dataset em duas partes: treinamento e teste
from sklearn.model_selection import train_test_split

## Montando um GridSearch
par_hls = [(100), (120), (150), (50,50), (60,60)]
par_activation = ['identity', 'logistic', 'relu', 'tanh']
par_solver = ['adam', 'sgd']
par_learning_rate_init = [0.0001, 0.001, 0.01]
par_max_iter = [500, 1000, 1500, 2000, 2500]

### Começando a contar o tempo
start_time = time.time()

for i in par_hls:
    for j in par_activation:
        for k in par_solver:
            for l in par_learning_rate_init:
                for m in par_max_iter:
                    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=19 )

                    mlp = MLPClassifier(hidden_layer_sizes=i, activation=j,
                                        solver=k, learning_rate_init=l, max_iter=m)

                    mlp.fit(X_train, y_train)
                    y_pred = mlp.predict(X_test)

                    accuracy = metrics.accuracy_score(y_test, y_pred)
                    print('Accuracy: %.3f, hidden: %s, activation = %s, solver = %s, learning = %s, max_iter = %s'
                     % (accuracy, i, j, k, l, m))

end_time = time.time()
# Tempo total de treinamento
execution_time = end_time - start_time
print(f"Tempo de execução para treinamento: {execution_time:.4f} segundos")
```

## Tarefa 2 — MLP nas duas bases
Mesma configuração das células acima (100 neurônios, relu, adam, `max_iter=500`), com validação cruzada 5-fold, aplicada ao BOW e ao TF-IDF.

```python
import warnings
import pandas as pd
from numpy import mean, std
from sklearn.neural_network import MLPClassifier
from sklearn.model_selection import KFold, cross_val_score

warnings.filterwarnings("ignore")

bases = {"BOW": "Imdb_BOW.csv", "TF-IDF": "Imdb_TF-IDF.csv"}
dados = {}
for nome, arquivo in bases.items():
    ds = pd.read_csv(arquivo, encoding="utf-8")
    dados[nome] = (ds.iloc[:, :-1], ds["label"])

kf = KFold(n_splits=5, random_state=1, shuffle=True)

for nome, (X, y) in dados.items():
    mlp = MLPClassifier(hidden_layer_sizes=(100), activation="relu", solver="adam",
                        max_iter=500, random_state=1)
    scores = cross_val_score(mlp, X, y, scoring="accuracy", cv=kf, n_jobs=-1)
    print("%-7s Accuracy: %.3f (%.3f)" % (nome, mean(scores), std(scores)))
```

## Tarefa 3 — Variando neurônios e iterações
Para cada base, testamos várias quantidades de neurônios na camada escondida e vários valores de `max_iter`, sempre com 5-fold. Também guardamos a acurácia no treino e quantas épocas o adam rodou de fato (`n_iter_`), já que ele para sozinho quando a perda deixa de melhorar.

```python
import time
from sklearn.model_selection import cross_validate

par_neuronios = [5, 10, 50, 100, 200, 500]
par_max_iter = [5, 10, 25, 50, 100, 200, 500, 1000]

resultados = []
inicio = time.time()

for nome, (X, y) in dados.items():
    for n in par_neuronios:
        for it in par_max_iter:
            mlp = MLPClassifier(hidden_layer_sizes=(n,), activation="relu", solver="adam",
                                max_iter=it, random_state=1)
            cv = cross_validate(mlp, X, y, scoring="accuracy", cv=kf, n_jobs=-1,
                                return_train_score=True, return_estimator=True)
            resultados.append({
                "Base": nome,
                "Neuronios": n,
                "Max_iter": it,
                "Acc_teste": cv["test_score"].mean(),
                "Desvio": cv["test_score"].std(),
                "Acc_treino": cv["train_score"].mean(),
                "Epocas": mean([e.n_iter_ for e in cv["estimator"]]),
            })

print(f"Tempo total: {time.time() - inicio:.1f} s")
df = pd.DataFrame(resultados)
df.to_csv("resultados_mlp.csv", index=False)
```

```python
pd.set_option("display.precision", 3)

for nome in bases:
    print(f"Acurácia de teste (5-fold) — {nome}  [linhas: neurônios, colunas: max_iter]")
    display(df[df.Base == nome].pivot(index="Neuronios", columns="Max_iter", values="Acc_teste"))
```

```python
print("Épocas realmente usadas pelo adam (média dos 5 folds) — BOW")
display(df[df.Base == "BOW"].pivot(index="Neuronios", columns="Max_iter", values="Epocas"))

print("Três melhores configurações de cada base")
display(df.sort_values("Acc_teste", ascending=False).groupby("Base").head(3))
```

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap
from matplotlib.ticker import FixedLocator, NullLocator, PercentFormatter

df = pd.read_csv("resultados_mlp.csv")
os.makedirs("graficos", exist_ok=True)

FUNDO, TINTA, TINTA2, MUDO = "#ffffff", "#0b0b0b", "#52514e", "#898781"
GRADE, EIXO = "#e1e0d9", "#c3c2b7"
AZUL, LARANJA = "#2a78d6", "#eb6834"
plt.rcParams.update({
    "figure.facecolor": FUNDO, "axes.facecolor": FUNDO, "savefig.facecolor": FUNDO,
    "axes.edgecolor": EIXO, "axes.linewidth": 0.8, "axes.labelcolor": TINTA2,
    "xtick.color": MUDO, "ytick.color": MUDO, "xtick.labelcolor": TINTA2, "ytick.labelcolor": TINTA2,
    "text.color": TINTA, "axes.spines.top": False, "axes.spines.right": False,
    "axes.grid": True, "grid.color": GRADE, "grid.linewidth": 0.8, "axes.axisbelow": True,
    "font.size": 11, "axes.titlesize": 12.5, "axes.titleweight": "bold", "axes.titlelocation": "left",
    "axes.titlepad": 10, "xtick.major.size": 0, "ytick.major.size": 0, "legend.frameon": False,
})
ITERS = [5, 10, 25, 50, 100, 200, 500, 1000]
BASES = ["BOW", "TF-IDF"]

def eixo_iteracoes(ax):
    ax.set_xscale("log")
    ax.xaxis.set_major_locator(FixedLocator(ITERS))
    ax.xaxis.set_minor_locator(NullLocator())
    ax.set_xticklabels([str(i) for i in ITERS])
    ax.set_xlim(4, 1300)
    ax.grid(axis="x", visible=False)
    ax.set_xlabel("max_iter (iterações de treino, escala log)")

def titulo(fig, texto, sub):
    fig.text(0.012, 0.975, texto, fontsize=15, fontweight="bold", va="top")
    fig.text(0.012, 0.915, sub, fontsize=11, color=TINTA2, va="top")

# 1) Acurácia de teste x iterações, uma linha por quantidade de neurônios
neur = {5: "#86b6ef", 50: "#2a78d6", 500: "#104281"}
fig, axes = plt.subplots(1, 2, figsize=(11, 4.9), sharey=True)
for ax, base in zip(axes, BASES):
    ax.axhline(0.5, color=EIXO, lw=1, zorder=1)
    for n, cor in neur.items():
        d = df[(df.Base == base) & (df.Neuronios == n)].sort_values("Max_iter")
        ax.plot(d.Max_iter, d.Acc_teste, color=cor, lw=2, marker="o", ms=6.5,
                mec=FUNDO, mew=1.5, solid_capstyle="round", solid_joinstyle="round",
                label=f"{n} neurônios", zorder=3)
    ax.set_title(base)
    eixo_iteracoes(ax)
    ax.yaxis.set_major_formatter(PercentFormatter(1.0, decimals=0))
    ax.set_ylim(0.45, 0.82)
axes[0].set_ylabel("Acurácia de teste (5-fold)")
axes[0].text(300, 0.507, "50% = chute", color=MUDO, fontsize=10, ha="center", va="bottom")
axes[0].annotate("acima de ~400 iterações\no adam já parou sozinho",
                 xy=(700, 0.752), xytext=(260, 0.62), color=TINTA2, fontsize=10, ha="center",
                 arrowprops=dict(arrowstyle="-", color=MUDO, lw=0.8))
titulo(fig, "Acurácia de teste conforme as iterações de treino",
       "MLP com uma camada escondida (relu, adam). 10 e 100–200 neurônios seguem o mesmo padrão; tabela completa no relatório.")
fig.legend(*axes[0].get_legend_handles_labels(), loc="upper right", bbox_to_anchor=(0.99, 0.99),
           ncol=3, fontsize=10.5, handlelength=1.6, columnspacing=1.4)
fig.tight_layout(rect=(0, 0, 1, 0.86))
fig.savefig("graficos/acuracia_vs_iteracoes.png", dpi=200)
plt.show()

# 2) Treino x teste: por que treinar mais piora
n = 10
fig, axes = plt.subplots(1, 2, figsize=(11, 4.9), sharey=True)
for ax, base in zip(axes, BASES):
    d = df[(df.Base == base) & (df.Neuronios == n)].sort_values("Max_iter")
    ax.fill_between(d.Max_iter, d.Acc_teste, d.Acc_treino, color=MUDO, alpha=0.10, lw=0, zorder=1)
    for col, cor, nome in [("Acc_treino", LARANJA, "Treino"), ("Acc_teste", AZUL, "Teste")]:
        ax.plot(d.Max_iter, d[col], color=cor, lw=2, marker="o", ms=6.5, mec=FUNDO, mew=1.5,
                solid_capstyle="round", solid_joinstyle="round", label=nome, zorder=3)
        ax.text(1150, d[col].iloc[-1], nome.lower(), color=TINTA2, fontsize=10, va="center")
    ax.set_title(base)
    eixo_iteracoes(ax)
    ax.set_xlim(4, 2000)
    ax.yaxis.set_major_formatter(PercentFormatter(1.0, decimals=0))
    ax.set_ylim(0.45, 1.02)
    meio = d[d.Max_iter == 500].iloc[0]
    ax.text(500, (meio.Acc_treino + meio.Acc_teste) / 2, "diferença =\noverfitting",
            color=TINTA2, fontsize=10, ha="center", va="center")
axes[0].set_ylabel("Acurácia (5-fold)")
titulo(fig, "Treino × teste com 10 neurônios",
       "Primeiro a rede não aprende (50%), depois aprende, e por fim decora o treino enquanto o teste cai.")
fig.legend(*axes[0].get_legend_handles_labels(), loc="upper right", bbox_to_anchor=(0.99, 0.99),
           ncol=2, fontsize=10.5, handlelength=1.6)
fig.tight_layout(rect=(0, 0, 1, 0.86))
fig.savefig("graficos/treino_vs_teste.png", dpi=200)
plt.show()

# 3) Mapa de calor com todas as combinações
azuis = ["#cde2fb", "#9ec5f4", "#6da7ec", "#3987e5", "#256abf", "#184f95", "#0d366b"]
cmap = LinearSegmentedColormap.from_list("azul", azuis)
fig, axes = plt.subplots(1, 2, figsize=(11, 4.6))
for ax, base in zip(axes, BASES):
    t = df[df.Base == base].pivot(index="Neuronios", columns="Max_iter", values="Acc_teste")
    m = ax.pcolormesh(t.values, cmap=cmap, vmin=0.5, vmax=0.79, edgecolors=FUNDO, linewidth=2)
    ax.invert_yaxis()
    ax.set_xticks([i + 0.5 for i in range(len(t.columns))], [str(c) for c in t.columns])
    ax.set_yticks([i + 0.5 for i in range(len(t.index))], [str(r) for r in t.index])
    ax.grid(False)
    for s in ax.spines.values():
        s.set_visible(False)
    ax.set_title(base)
    ax.set_xlabel("max_iter")
    r, c = divmod(int(t.values.argmax()), t.shape[1])
    ax.text(c + 0.5, r + 0.5, f"{t.values[r, c]:.0%}".replace(".", ","), ha="center", va="center",
            color="#ffffff", fontsize=10, fontweight="bold")
axes[0].set_ylabel("Neurônios")
fig.subplots_adjust(left=0.07, right=0.86, top=0.8, bottom=0.13, wspace=0.18)
cax = fig.add_axes([0.885, 0.13, 0.014, 0.67])
cb = fig.colorbar(m, cax=cax, format=PercentFormatter(1.0, decimals=0))
cb.outline.set_visible(False)
cb.ax.tick_params(colors=MUDO, labelcolor=TINTA2, size=0)
cb.set_label("Acurácia de teste (5-fold)", color=TINTA2)
fig.text(0.012, 0.975, "Todas as combinações de neurônios × iterações", fontsize=15,
         fontweight="bold", va="top")
fig.text(0.012, 0.915, "Mais escuro = maior acurácia. O número marca a melhor combinação de cada base.",
         fontsize=11, color=TINTA2, va="top")
fig.savefig("graficos/mapa_neuronios_iteracoes.png", dpi=200)
plt.show()
```
